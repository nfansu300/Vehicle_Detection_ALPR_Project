```python
from ultralytics import YOLO
import os

# ============================================================
# CONFIGURATION
# ============================================================

# Fine-tuned YOLOv8 model
MODEL_PATH = "/content/drive/MyDrive/Stage_PFA/experiments/Exp05_YOLO11N_EPOCHS_100/weights/best.pt"

# Dataset configuration
DATA_YAML = "/content/Final-Complete-Dataset-4/data.yaml"

# Output directory
PROJECT_DIR = "/content/drive/MyDrive/Stage_PFA/test_evaluation"
EXPERIMENT_NAME = "Exp05_YOLO11N_EPOCHS_100_TEST"

# ============================================================
# LOAD MODEL
# ============================================================

print("=" * 70)
print("YOLOv8 TEST EVALUATION")
print("=" * 70)

print(f"\nLoading model:")
print(MODEL_PATH)

model = YOLO(MODEL_PATH)

print("\nModel loaded successfully!")
print(f"Model: {MODEL_PATH}")

# ============================================================
# TEST EVALUATION
# ============================================================

print("\n" + "=" * 70)
print("RUNNING TEST EVALUATION")
print("=" * 70)

results = model.val(
    data=DATA_YAML,
    split="test",

    # Image size used during training
    imgsz=640,

    # Batch size
    batch=32,

    # GPU
    device=0,

    # Save evaluation plots
    plots=True,

    # Save JSON results
    save_json=True,

    # Verbose output
    verbose=True,

    # Output directory
    project=PROJECT_DIR,
    name=EXPERIMENT_NAME,
    exist_ok=True
)

# ============================================================
# DISPLAY RESULTS
# ============================================================

print("\n" + "=" * 70)
print("TEST EVALUATION RESULTS")
print("=" * 70)

print(f"\nPrecision (P):      {results.box.mp:.4f}")
print(f"Recall (R):         {results.box.mr:.4f}")
print(f"mAP@50:             {results.box.map50:.4f}")
print(f"mAP@50-95:          {results.box.map:.4f}")

print("\n" + "=" * 70)
print("PER-CLASS RESULTS")
print("=" * 70)

for i, class_name in model.names.items():
    print(
        f"{class_name:20s} "
        f"P={results.box.p[i]:.4f} "
        f"R={results.box.r[i]:.4f} "
        f"mAP50={results.box.ap50[i]:.4f} "
        f"mAP50-95={results.box.ap[i]:.4f}"
    )

print("\n" + "=" * 70)
print("TEST EVALUATION COMPLETED")
print("=" * 70)

print(f"\nResults saved to:")
print(f"{PROJECT_DIR}/{EXPERIMENT_NAME}")
```
